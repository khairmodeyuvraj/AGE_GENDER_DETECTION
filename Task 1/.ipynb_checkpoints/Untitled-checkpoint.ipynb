```python
import os
import cv2
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split
from sklearn.metrics import confusion_matrix

import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Conv2D, MaxPooling2D, Flatten, Dense, Dropout
import warnings
warnings.filterwarnings('ignore')
```

```python
# PATH SETUP

# Current folder → Task 1
CURRENT_DIR = os.getcwd()

# Go one level UP → project root
BASE_DIR = os.path.dirname(CURRENT_DIR)

# Dataset location
DATASET_PATH = os.path.join(BASE_DIR, "Datasets", "UTKFace")

print("Current Directory :", CURRENT_DIR)
print("Project Root      :", BASE_DIR)
print("Dataset Path      :", DATASET_PATH)
print("Dataset Exists    :", os.path.exists(DATASET_PATH))

IMG_SIZE = 128
MAX_IMAGES = 4000

images = []
ages = []
genders = []
hair_labels = []

count = 0

# DATA LOADING
for img_name in os.listdir(DATASET_PATH):

    try:
        age = int(img_name.split("_")[0])
        gender = int(img_name.split("_")[1])

        img_path = os.path.join(DATASET_PATH, img_name)

        img = cv2.imread(img_path)
        if img is None:
            continue

        img = cv2.resize(img, (IMG_SIZE, IMG_SIZE))
        img = np.asarray(img, dtype=np.float32) / 255.0

        # Hair proxy logic
        hair = 1 if (gender == 0 and age >= 15) else 0

        images.append(img)
        ages.append(age)
        genders.append(gender)
        hair_labels.append(hair)

        count += 1
        if count >= MAX_IMAGES:
            break

    except:
        continue

# Convert to numpy
X = np.array(images, dtype=np.float32)
y_age = np.array(ages)
y_gender = np.array(genders)
y_hair = np.array(hair_labels)

print("\n✅ Dataset Loaded Successfully")
print("Images Shape :", X.shape)
```

**train_test_split**

```python
X_train, X_test, age_train, age_test, gender_train, gender_test, hair_train, hair_test = train_test_split(
    X, y_age, y_gender, y_hair,
    test_size=0.2,
    random_state=42
)
```

```python
MODELS_DIR = os.path.join(BASE_DIR, "models")
os.makedirs(MODELS_DIR, exist_ok=True)

print("Models will be saved in:", MODELS_DIR)
```

**Age model**

```python
age_model = Sequential([
    Conv2D(32,(3,3),activation='relu',input_shape=(128,128,3)),
    MaxPooling2D(),

    Conv2D(64,(3,3),activation='relu'),
    MaxPooling2D(),

    Conv2D(128,(3,3),activation='relu'),
    MaxPooling2D(),

    Flatten(),
    Dense(128,activation='relu'),
    Dropout(0.5),
    Dense(1)
])

age_model.compile(optimizer='adam', loss='mse', metrics=['mae'])
```

**Train Age**

```python
history_age = age_model.fit(
    X_train, age_train,
    validation_data=(X_test, age_test),
    epochs=10,
    batch_size=32
)
age_model.save(os.path.join(MODELS_DIR, "age_model.h5"))
```

**Gender model**

```python
gender_model = Sequential([
    Conv2D(32,(3,3),activation='relu',input_shape=(128,128,3)),
    MaxPooling2D(),

    Conv2D(64,(3,3),activation='relu'),
    MaxPooling2D(),

    Flatten(),
    Dense(128,activation='relu'),
    Dropout(0.5),
    Dense(1,activation='sigmoid')
])

gender_model.compile(
    optimizer='adam',
    loss='binary_crossentropy',
    metrics=['accuracy']
)
```

**Train Gender model**

```python
history_gender = gender_model.fit(
    X_train, gender_train,
    validation_data=(X_test, gender_test),
    epochs=10,
    batch_size=32
)

gender_model.save(os.path.join(MODELS_DIR, "gender_model.h5"))
```

**Hair model**

```python
hair_model = Sequential([
    Conv2D(32,(3,3),activation='relu',input_shape=(128,128,3)),
    MaxPooling2D(),

    Conv2D(64,(3,3),activation='relu'),
    MaxPooling2D(),

    Flatten(),
    Dense(128,activation='relu'),
    Dropout(0.5),
    Dense(1,activation='sigmoid')
])

hair_model.compile(
    optimizer='adam',
    loss='binary_crossentropy',
    metrics=['accuracy']
)
```

**Train Hair model**

```python
history_hair = hair_model.fit(
    X_train, hair_train,
    validation_data=(X_test, hair_test),
    epochs=10,
    batch_size=32
)
hair_model.save(os.path.join(MODELS_DIR, "hair_model.h5"))
```

**Visualization**

```python
plt.figure()

plt.plot(history_age.history['mae'])
plt.plot(history_age.history['val_mae'])

plt.title("Age Model Mean Absolute Error")
plt.xlabel("Epoch")
plt.ylabel("MAE")
plt.legend(['Train', 'Validation'])

plt.show()
```

```python
plt.plot(history_gender.history['accuracy'])
plt.plot(history_gender.history['val_accuracy'])
plt.title("Gender Model Accuracy")
plt.xlabel("Epoch")
plt.ylabel("Accuracy")
plt.legend(['Train', 'Validation'])
plt.show()
```

```python
plt.figure()

plt.plot(history_hair.history['accuracy'])
plt.plot(history_hair.history['val_accuracy'])

plt.title("Hair Length Model Accuracy")
plt.xlabel("Epoch")
plt.ylabel("Accuracy")
plt.legend(['Train', 'Validation'])

plt.show()
```

**Confusion Matrix**

```python
preds = (gender_model.predict(X_test) > 0.5).astype(int)
cm = confusion_matrix(gender_test, preds)

sns.heatmap(cm, annot=True, fmt='d')
plt.title("Gender Confusion Matrix")
plt.show()
```
